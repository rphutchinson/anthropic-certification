```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
# Create API Client
from anthropic import Anthropic

client = Anthropic()
model = "anthropic.claude-4-6-sonnet"
```

```python
# Make Request
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role":"user",
            "content":"What is quantum computing? Answer in one sentence."
        }
    ]
)
```
