```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()

# Create API Client
from anthropic import Anthropic

client = Anthropic()
model = "anthropic.claude-4-6-sonnet"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model":model,
        "max_tokens": 1000,
        "messages": messages
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    
    return message.content[0].text
```

```python
system = """
You are a senior Python developer. 
Your job is to respond as conciesly as possible to requests for code creation. 
Your responses should include only the code itself with no explanations or other superfluous information. 
Do not give the user options for different implementations, use your judgement to select the best solution and return only the code for that solution
"""

messages = []
add_user_message(messages, "Write the Python code that checks a string for duplicate characters")
answer = chat(messages, system)

answer


```
